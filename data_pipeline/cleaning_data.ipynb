```python
# Task 2 cleaning books data set using pandas
import pandas as pd

DEFAULT_GBP_VALUE = 105.50
```

```python
#Load data set using padas
df = pd.read_csv('books_dataset.csv')

#Strip the currency symbol from price and convert it to a float column price_gbp.
df["price_gbp"] = df["price_gbp"].str.replace("£", "", regex=False).astype(float)

#Convert the text star rating (One…Five) into an integer column rating (1–5).
RATING_MAP = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}
df['rating'] = df['rating'].map(RATING_MAP)
```

```python
#Price convertion
df["price_inr"] = (df["price_gbp"] * DEFAULT_GBP_VALUE).round(2)
```

```python
df.to_csv("cleaned_books_dataset.csv",index=False)
```
